```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
from bs4 import BeautifulSoup
import time
import re
```

```python
driver = webdriver.Firefox()
url = 'https://www.carsome.id/beli-mobil-bekas'
driver.get(url)
soup_list = []

while True:
    try:
        next = WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.XPATH, "//button[@class='v-pagination__navigation' and @aria-label='Next page']")))
        if 'a-disabled' in next.get_attribute('class'):
            break
        next.click()
    except Exception as e:
        print(f"An error occurred: {e}")
        break

    time.sleep(5)  # Add a delay to allow the page to load
    soup = BeautifulSoup(driver.page_source, 'html.parser')
    soup_list.append(soup)

# Close the browser when done
driver.quit()
```

```python
car = []

for html in soup_list:
    links = html.find_all("div", class_="mod-card__footer")
    # Loop through the list of links
    for link in links:
        car_detail = {}  # Create a new car_detail dictionary for each car
        
        # Extract the text from all <p> elements within the link
        title_parts = link.find_all("p")
        title = " ".join(part.get_text(strip=True) for part in title_parts)
        car_detail['name'] = re.sub(r'\s+', ' ', title)

        # Search for details within the context of the specific car listing (using 'link')
        details1 = link.find("div", class_="mod-card__car-other")
        car_detail1 = [span.get_text(strip=True) for span in details1.find_all("span")]
        car_detail['detail1'] = car_detail1

        details2 = link.find("div", class_="mod-card__high-tag-list")
        car_detail2 = [span.get_text(strip=True) for span in details2.find_all("span")]
        car_detail['detail2'] = car_detail2

        price = link.find("div", class_="mod-card__price__total")
        price = price.text.strip()
        car_detail['price'] = re.sub(r'\s+', ' ', price)

        price = link.find("div", class_="mod-card__price")
        price = price.text.strip()
        car_detail['price'] = re.sub(r'\s+', ' ', price)

        price_m = link.find("div", class_="mod-card__price")
        price_m = "".join(price_m.get_text(strip=True) for price_m in price_m.find_all("span"))
        car_detail['monthly'] = re.sub(r'\s+', ' ', price_m)

        # Append the car_detail dictionary to the car list
        car.append(car_detail)

    # Print or use the 'car' list containing all car details
    print(price_m)
```
