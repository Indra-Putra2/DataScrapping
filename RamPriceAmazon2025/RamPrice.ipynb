```python
from io import StringIO
import pandas as pd
import re
import numpy as np
```

```python
# Get data without the bad one
df = pd.read_csv("RamData.csv",
    on_bad_lines="skip"
)

print(df.shape)
```

```python
df
```

```python
#Bad Data Isolation
bad_rows = []

with open("RamData.csv") as f:
    header = f.readline()  # skip header
    for i, line in enumerate(f, start=1):
        if line.count(",") != 9:
            bad_rows.append(line)

len(bad_rows)
```

```python
for bad_row in bad_rows:
    print(bad_row)
```

```python
# Fixing the bad data using RegEx
fixed = []
pattern = r"^(.*?)([A-Z]{3}\d.*)"

for bad_row in bad_rows:
    match = re.search(pattern, bad_row)
    fixedName = match.group(1).replace(",","") + ","
    fixedCombined = fixedName + match.group(2)
    fixed.append(fixedCombined)
    print(fixedCombined)
```

```python
fixed_df = pd.read_csv(
    StringIO("\n".join(fixed)),
    header=None
)

fixed_df.columns = df.columns

fixed_df
```

```python
df = pd.concat([df, fixed_df], ignore_index=True)

print(df.shape)
```

```python
df['Brand'] = df['Name'].apply(lambda x: x.split()[0]).astype('string')
```

```python
df['Brand'].unique()
```

```python
df['Brand'] = df['Brand'].replace("Super", "Super Talent")
df['Brand'] = df['Brand'].replace("Neo", "Neo Forza")
df['Brand'] = df['Brand'].replace("SK", "SK Hynix")
```

```python
df['Brand'].unique()
```

```python
df[['Memory Type', 'Memory Speed']] = df['Speed'].str.split("-", expand=True)
df['Memory Type'] = df['Memory Type'].astype('string')
df['Memory Speed'] = df['Memory Speed'].astype(int)
```

```python
df['First Word Latency'] = df['First Word Latency'].str.extract(r"(.+\d)").fillna(0).astype(float)
df['Price']= df['Price'].str.extract(r'(\d+\.\d+)|(\d+)')[0].fillna(0).astype(float)
df['Color'] = df['Color'].fillna("None")
df["Color"] = df["Color"].str.split("/").apply(lambda x: [c.strip() for c in x])
df['Module'] = df['Module'].astype('string')
df['Name'] = df['Name'].astype('string')

extracted = df["Module"].str.extract(r"(\d+)\s*x\s*(\d+)(\w+)")
df["Total Capacity"] = (
    extracted[0].astype(int)
    * extracted[1].astype(int)
    / extracted[2].str.lower().map({"gb": 1, "mb": 1024})
).astype(float)
df["Price Per GB"] = (df["Price"] / df["Total Capacity"]).round(2)
```

```python
df = df.drop(columns=['Speed'])
```

```python
new_order = ['Name', 'Brand', 'Memory Type', 'Memory Speed', 'Module', 'Total Capacity', 'First Word Latency', 'CAS Latency', 'Price', 'Price Per GB', 'Color', 'Rating', 'Review Count']
df = df[new_order]
df
```

```python
brands = df["Brand"].unique()

brands_df = pd.Series(brands, name="Name").to_frame()

brands_df["Id"] = brands_df.index

brands_df = brands_df[["Id", "Name"]]

brands_df.to_csv("brands.csv", index=False)
```

```python
brand_map = dict(zip(brands_df["Name"], brands_df["Id"]))

df["BrandId"] = df["Brand"].map(brand_map)
```

```python
brand_index = df.columns.get_loc("Brand")

df["BrandId"] = df["Brand"].map(brand_map)
df.drop(columns=["Brand"], inplace=True)
df.insert(brand_index, "BrandId", df.pop("BrandId"))
```
